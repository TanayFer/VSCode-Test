```python
#-----Q9------=
import numpy as np

a1 = np.array([1, 2, 3, 4])
a2 = np.array([[1, 2], [3, 4]])
a3 = np.array([[[1, 2]], [[3, 4]]])

for name, a in [("1D", a1), ("2D", a2), ("3D", a3)]:
    print(f"{name} = Dim: {a.ndim}, Shape: {a.shape}, Size: {a.size}")
```
