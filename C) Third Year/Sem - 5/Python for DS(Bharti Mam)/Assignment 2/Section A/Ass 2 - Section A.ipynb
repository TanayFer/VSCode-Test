```python
df_clean = df1.dropna()
print(df_clean)
```

```python
df_mean = df1.copy()
df_mean['Math'] = df_mean['Math'].fillna(df_mean['Math'].mean())

df_median = df1.copy()
df_median['Physics'] = df_median['Physics'].fillna(df_median['Physics'].median())

df_mode = df1.copy()
df_mode['Grade'] = df_mode['Grade'].fillna(df_mode['Grade'].mode())

df_clean = df1.copy()

df_clean['Math'] = df_clean['Math'].fillna(df_clean['Math'].mean())
df_clean['Physics'] = df_clean['Physics'].fillna(df_clean['Physics'].median())
df_clean['Chemistry'] = df_clean['Chemistry'].fillna(df_clean['Chemistry'].mean())

df_clean['Grade'] = df_clean['Grade'].fillna(df_clean['Grade'].mode())
df_clean['Comment'] = df_clean['Comment'].fillna(df_clean['Comment'].mode())

print("Original DataFrame:")
print(df1)
print("\nCleaned DataFrame:")
print(df_clean)
```
