```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
file = "Student_Performance.xlsx"
df = pd.read_excel(file)
print(df)
```

```python
print("\nShape of Dataset: ",df.shape)
```

```python
print(df.head())
print(df.tail())
print(df.info())
print(df.describe())
```

```python
print("Missing values:")
print(df.isnull().sum())

df_removed = df.dropna()
print("\nAfter removing missing values:")
print(df_removed)

df_mean = df.copy()

for col in df_mean.select_dtypes("number"):
    df_mean[col] = df_mean[col].fillna(df_mean[col].mean())

print("\nAfter filling with mean:")
print(df_mean)

df_filled = df.copy()

for col in df_filled.select_dtypes("number"):
    df_filled[col] = df_filled[col].fillna(df_filled[col].median())

for col in df_filled.select_dtypes("object"):
    df_filled[col] = df_filled[col].fillna(df_filled[col].mode()[0])

print("\nAfter filling with median and mode:")
print(df_filled)
```

```python
print("Data types before conversion:")
print(df.dtypes)

df["StudentID"] = df["StudentID"].astype(int)
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
df["Attendance"] = df["Attendance"].astype(float)

print("\nData types after conversion:")
print(df.dtypes)

print("\nUpdated DataFrame:")
print(df)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day

df["Day_Name"] = df["Date"].dt.day_name()

print(df)
```

```python
scaler = MinMaxScaler()

df[["Marks", "Attendance"]] = scaler.fit_transform(
    df[["Marks", "Attendance"]]
)

print(df[["Marks", "Attendance"]])
```

```python
scaler = StandardScaler()

df[['Marks', 'Attendance']] = scaler.fit_transform(
    df[['Marks', 'Attendance']]
)

print("Standardized Values:")
print(df)
```

```python
encoder = LabelEncoder()

df["Result_Encoded"] = encoder.fit_transform(df["Result"])

print(df[["Result", "Result_Encoded"]])
```

```python
df = pd.get_dummies(df, columns=["Department"])
print(df)
```
