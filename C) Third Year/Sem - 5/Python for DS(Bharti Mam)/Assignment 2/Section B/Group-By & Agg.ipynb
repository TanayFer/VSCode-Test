```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import LabelEncoder
```

```python
file = "Student_Performance.xlsx"
df = pd.read_excel(file)
print(df)
```

```python
average_marks = df.groupby("Department")["Marks"].mean()
print(average_marks)
```

```python
df.columns = df.columns.str.strip()

result = df.groupby("Department")["Marks"].agg(
    ["mean", "max", "min", "count"]
)

print(result)
```

```python
print("Original Data:")
print(df)

print("\nUsing loc[]:")
print(df.loc[:, ["Name", "Department", "Marks"]])

print("\nUsing iloc[]:")
print(df.iloc[:, [1, 2, 5]])

print("\nCondition 1: Marks Greater Than 85")
print(df.loc[df["Marks"] > 85, ["Name", "Department", "Marks"]])

print("\nCondition 2: Department is CS")
print(df.loc[df["Department"] == "CS", ["Name", "Department", "Marks"]])
```

```python
df["Date"] = pd.to_datetime(df["Date"])

result = df.groupby(df["Date"].dt.to_period("M"))["Marks"].agg(["sum", "mean"])

print(result)
```
