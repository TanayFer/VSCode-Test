```python
#--------Q2--------
details = pd.read_excel("student_details.xlsx")
performance = pd.read_excel("student_performance.xlsx")

#Inner Join
inner_join = pd.merge(details, performance, on="Student_ID", how="inner")

#Left Join
left_join = pd.merge(details, performance, on="Student_ID", how="left")

#O/P
print("Inner Join:")
display(inner_join)

print("Left Join:")
display(left_join)

#Comparing
print("Number of records in Inner Join:", len(inner_join))
print("Number of records in Left Join:", len(left_join))
```

```python
#-------Q3---------
details = pd.read_excel("student_details.xlsx")
performance = pd.read_excel("student_performance.xlsx")

#Right Join
right_join = pd.merge(details, performance, on="Student_ID", how="right")

#Outer Join
outer_join = pd.merge(details, performance, on="Student_ID", how="outer")

#O/P
print("Right Join:")
display(right_join)

print("Outer Join:")
display(outer_join)

#Records of NaN
print("NaN records : Right Join:")
display(right_join[right_join.isna().any(axis=1)])

```

```python
#---------Q4--------
details = pd.read_excel("student_details.xlsx")
performance = pd.read_excel("student_performance.xlsx")

#Perform joins
inner = pd.merge(details, performance, on="Student_ID", how="inner")
left = pd.merge(details, performance, on="Student_ID", how="left")
right = pd.merge(details, performance, on="Student_ID", how="right")
outer = pd.merge(details, performance, on="Student_ID", how="outer")

#O/P
print("INNER JOIN")
display(inner)

print("LEFT JOIN")
display(left)

print("RIGHT JOIN")
display(right)

print("OUTER JOIN")
display(outer)

#Calculate records and missing values
results = pd.DataFrame({
    "Join": ["Inner", "Left", "Right", "Outer"],
    "Records": [len(inner), len(left), len(right), len(outer)],
    "Missing Values": [
        inner.isna().sum().sum(),
        left.isna().sum().sum(),
        right.isna().sum().sum(),
        outer.isna().sum().sum()
    ]
})

print("Join Comparison:")
display(results)
```

```python
#-------Q5------
df = pd.read_excel("student_details.xlsx")

#Divide in 2
part1 = df.iloc[:len(df)//2]
part2 = df.iloc[len(df)//2:]

#Combine
combined = pd.concat([part1, part2])

#O/P
print("First Part:")
display(part1)

print("Second Part:")
display(part2)

print("Combined DataFrame:")
display(combined)
```

```python
#--------Q6-----------
df = pd.read_excel("student_details.xlsx")

#Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

#Memory before conversion
before = df.memory_usage(deep=True).sum()

#Convert
df["Student_ID"] = df["Student_ID"].astype("int16")
df["Year"] = df["Year"].astype("int8")

#Memory after conversion
after = df.memory_usage(deep=True).sum()

#Calculate reduction
reduction = ((before - after) / before) * 100

print("Before:", before, "bytes")
print("After:", after, "bytes")
print("Reduction:", round(reduction, 2), "%")
```
