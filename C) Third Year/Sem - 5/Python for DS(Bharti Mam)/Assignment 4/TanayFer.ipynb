```python
#Importing 
import pandas as pd
```

```python
df = pd.read_excel("Stud_Data.xlsx")
print(df.head())
```

```python
#Q1 
#Write a Python program using groupby() and a 
#lambda function to find the difference between the 
#maximum and minimum Total_Marks for each School_Name.
result1 = df.groupby("School_Name")["Total_Marks"].apply(lambda x: x.max() - x.min())
print("Diffrence BetN Max and Min Total marks")
print(result1)    
```

```python
#2.	Write a Python program using groupby() 
#and a lambda function to calculate the average Percentage for each City.
result2 = df.groupby("City")["Percentage"].apply(lambda x: x.mean())
print("Average City Percentage")
print(result2)
```

```python
#Q3 Write a Python program using .apply() 
#and a lambda function to create a new column named Result. 
#Set the value to "Pass" if Total_Marks is greater than or 
#equal to 120; otherwise, set it to "Fail".
df["Result"] = df["Total_Marks"].apply(lambda x: "Pass" if x >= 120 else "Fail")
print("\n3. Result:")
print(df[["Student_Name","School_Name","Total_Marks", "Result"]])
```

```python
#4. Write a Python program using groupby() and .transform() to create a new column 
#named School_Avg_Marks containing the average Total_Marks of the respective 
#School_Name for each student.
df["School_Avg_Marks"] = df.groupby("School_Name")["Total_Marks"].transform("mean")
print("\n4. School Average Marks:")
print(df[["Student_Name", "School_Name", "Total_Marks", "School_Avg_Marks"]])
```

```python
#Q5
df["Performance"] = df["Percentage"].apply(
    lambda x: "Excellent" if x >= 80
    else "Good" if x >= 60
    else "Needs Improvement"
)
df["City_Avg_Percentage"] = df.groupby("City")["Percentage"].transform("mean")

print(df[[
    "Student_Name",
    "City",
    "Percentage",
    "Performance",
    "City_Avg_Percentage"
]])
```

```python
#Q6
import pandas as pd
def remove_duplicates(df):
    return df.drop_duplicates()

def clean_total_marks(df):
    df["Total_Marks"] = pd.to_numeric(df["Total_Marks"], errors="coerce")
    return df

def fill_missing_total_marks(df):
    df["Total_Marks"] = df["Total_Marks"].fillna(df["Total_Marks"].mean())
    return df

def calculate_result(df):
    df["Result"] = df["Total_Marks"].apply(
        lambda x: "Pass" if x >= 40 else "Fail"
    )
    return df

df = (
    df
    .pipe(remove_duplicates)
    .pipe(clean_total_marks)
    .pipe(fill_missing_total_marks)
    .pipe(calculate_result)
)
print(df)
```

```python
#Q7
import pandas as pd
def remove_duplicates(df):
    return df.drop_duplicates()
def clean_total_marks(df):
    df["Total_Marks"] = pd.to_numeric(df["Total_Marks"], errors="coerce")
    return df
def fill_missing_total_marks(df):
    df["Total_Marks"] = df["Total_Marks"].fillna(df["Total_Marks"].mean())
    return df
def calculate_result(df):
    df["Result"] = df["Total_Marks"].apply(
        lambda x: "Pass" if x >= 40 else "Fail"
    )
    return df
try:
    df = (
        df
        .pipe(remove_duplicates)
        .pipe(clean_total_marks)
        .pipe(fill_missing_total_marks)
        .pipe(calculate_result)
    )
    df.to_csv("cleaned_students.csv", index=False)
    print("Pipeline executed successfully.")
    print("Cleaned data saved as cleaned_students.csv")
except FileNotFoundError:
    print("Input file not found.")
except pd.errors.EmptyDataError:
    print("Input file is empty.")
except Exception as e:
    print("Unexpected error:", e)
```
