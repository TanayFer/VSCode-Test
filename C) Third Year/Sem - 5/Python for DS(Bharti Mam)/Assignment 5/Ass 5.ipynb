```python
import pandas as pd
from scipy import stats
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("property_data_2000.csv")
```

```python
area = df["Area_sqft"].dropna()
result = stats.describe(area)
print("Count:",result.nobs)
print("Minimum:",result.minmax[0])
print("Maximum:",result.minmax[1])
print("Mean:",result.mean)
print("Variance:",result.variance)
```

```python
prices = df["Price_Lakhs"].dropna()
mode_result = stats.mode(prices,keepdims=False)
standard_deviation = stats.tstd(prices)
print("Mode of Price_Lakhs:",mode_result.mode)
print("Mode Frequency: ",mode_result.count)
print("Standard Deviation:",standard_deviation)
```

```python
df = df.dropna(subset=["Area_sqft","Price_Lakhs"])
X = df[["Area_sqft"]]
y = df[["Price_Lakhs"]]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)
print("Totoal Records:",len(df))
print("Training Records:",len(X_train))
print("testing Records:",len(X_test))
```
